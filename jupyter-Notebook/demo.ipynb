```python
"""PyCharm (No LangChain) — OpenRouter via OpenAI SDK (minimal, our style)"""

# ============================================================================
# IMPORTS
# ============================================================================
import os  # จัดการตัวแปรสภาพแวดล้อม
from dotenv import load_dotenv  # โหลดค่าใน .env
from openai import OpenAI  # ใช้ OpenAI SDK กับ OpenRouter ผ่าน base_url

# ============================================================================
# ENVIRONMENT SETUP (load once at module level, fail-fast)
# ============================================================================
load_dotenv()  # โหลด .env ทันทีเมื่อรันไฟล์
OPENROUTER_API_KEY  = os.getenv("OPENROUTER_API_KEY")
OPENROUTER_BASE_URL = os.getenv("OPENROUTER_BASE_URL")
if not OPENROUTER_API_KEY:
    raise RuntimeError("Missing OPENROUTER_API_KEY in .env")
if not OPENROUTER_BASE_URL:
    raise RuntimeError("Missing OPENROUTER_BASE_URL in .env")

# ============================================================================
# MAIN
# ============================================================================
def main():
    # สร้าง client ชี้ไปที่ OpenRouter (ต้องใส่ทั้ง api_key และ base_url)
    client = OpenAI(api_key=OPENROUTER_API_KEY, base_url=OPENROUTER_BASE_URL)

    # ตัวอย่างสั้น ๆ: ขอให้สรุป 1 ประโยค (คุณเปลี่ยนข้อความได้)
    resp = client.chat.completions.create(
        model="openai/gpt-4o",
        messages=[{"role": "user", "content": "สรุปหนึ่งประโยคว่า OpenRouter คืออะไร"}],
        temperature=0.7,
    )
    print(resp.choices[0].message.content)

# ============================================================================
# SCRIPT EXECUTION
# ============================================================================
if __name__ == "__main__":
    main()
```
