```python
"""PyCharm (No LangChain) — OpenRouter chat (replace Azure completions, our style)"""

# ============================================================================
# IMPORTS
# ============================================================================
import os
from dotenv import load_dotenv
from openai import OpenAI

# ============================================================================
# ENVIRONMENT SETUP (fail-fast)
# ============================================================================
load_dotenv()
OPENROUTER_API_KEY  = os.getenv("OPENROUTER_API_KEY")
OPENROUTER_BASE_URL = os.getenv("OPENROUTER_BASE_URL")
if not OPENROUTER_API_KEY:
    raise RuntimeError("Missing OPENROUTER_API_KEY in .env")
if not OPENROUTER_BASE_URL:
    raise RuntimeError("Missing OPENROUTER_BASE_URL in .env")

# ============================================================================
# MAIN
# ============================================================================
def main():
    client = OpenAI(api_key=OPENROUTER_API_KEY, base_url=OPENROUTER_BASE_URL)

    prompt = (
        "Suggest three names for a new pet salon business. "
        "The generated name ideas should evoke positive emotions and the following key features: "
        "Professional, friendly, Personalized Service."
    )

    resp = client.chat.completions.create(
        model="openai/gpt-4o",
        temperature=0.7,
        max_tokens=100,
        n=3,  # return 3 choices
        messages=[
            {"role": "system", "content": "You generate short, catchy brand names."},
            {"role": "user", "content": prompt},
        ],
    )

    print("Prompt:", prompt)
    for i, choice in enumerate(resp.choices, 1):
        print(f"Response {i}:", choice.message.content.strip())

# ============================================================================
# SCRIPT EXECUTION
# ============================================================================
if __name__ == "__main__":
    main()
```
