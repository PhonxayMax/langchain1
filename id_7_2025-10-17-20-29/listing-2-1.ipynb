```python
"""PyCharm (No LangChain) — tiktoken minimal counter (our simple style)"""

# ============================================================================
# IMPORTS
# ============================================================================
import tiktoken as tk  # token encoder

# ============================================================================
# FUNCTIONS
# ============================================================================
def count_tokens(text: str, encoding_name: str = "cl100k_base") -> int:
    enc = tk.get_encoding(encoding_name)
    return len(enc.encode(text))

# ============================================================================
# MAIN
# ============================================================================
def main():
    prompt = "I have a white dog named Champ"
    print("Number of tokens:", count_tokens(prompt, "cl100k_base"))
    # ถ้าใช้โมเดลตระกูล GPT-4o แนะนำ (ตัวเลือก): "o200k_base"
    # print("Tokens (o200k_base):", count_tokens(prompt, "o200k_base"))

if __name__ == "__main__":
    main()
```
