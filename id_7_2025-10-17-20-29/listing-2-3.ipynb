```python
"""PyCharm (No LangChain) — NagaAI Embeddings via OpenAI SDK (minimal, our style)"""

# ============================================================================
# IMPORTS
# ============================================================================
import os  # ตัวแปรสภาพแวดล้อม
from dotenv import load_dotenv  # โหลดค่าใน .env
from openai import OpenAI  # ใช้ OpenAI SDK ชี้ไปที่ NagaAI base_url

# ============================================================================
# ENVIRONMENT SETUP (load once at module level, fail-fast)
# ============================================================================
load_dotenv()
NAGA_API_KEY  = os.getenv("NAGA_API_KEY")
NAGA_BASE_URL = os.getenv("NAGA_BASE_URL")  # ควรเป็น https://api.naga.ac/v1
if not NAGA_API_KEY:
    raise RuntimeError("Missing NAGA_API_KEY in .env")
if not NAGA_BASE_URL:
    raise RuntimeError("Missing NAGA_BASE_URL in .env")

# ============================================================================
# FUNCTIONS
# ============================================================================
def get_embedding(text: str) -> list[float]:
    """ขอเวกเตอร์ embedding จาก NagaAI (model: text-embedding-3-small)"""
    client = OpenAI(api_key=NAGA_API_KEY, base_url=NAGA_BASE_URL)
    resp = client.embeddings.create(model="text-embedding-3-small", input=text)
    return resp.data[0].embedding

# ============================================================================
# MAIN
# ============================================================================
def main():
    text = "I have a white dog named Champ."
    emb = get_embedding(text)
    print("Embedding Length:", len(emb))
    print("Embedding (first 5):", emb[:5])

# ============================================================================
# SCRIPT EXECUTION
# ============================================================================
if __name__ == "__main__":
    main()
```
