```python
"""PyCharm (No LangChain) — OpenRouter chat (multiple responses approach)"""

# ============================================================================
# IMPORTS
# ============================================================================
import os
from dotenv import load_dotenv
from openai import OpenAI

# ============================================================================
# ENVIRONMENT SETUP (fail-fast)
# ============================================================================
load_dotenv()
OPENROUTER_API_KEY  = os.getenv("OPENROUTER_API_KEY")
OPENROUTER_BASE_URL = os.getenv("OPENROUTER_BASE_URL")
if not OPENROUTER_API_KEY:
    raise RuntimeError("Missing OPENROUTER_API_KEY in .env")
if not OPENROUTER_BASE_URL:
    raise RuntimeError("Missing OPENROUTER_BASE_URL in .env")

# ============================================================================
# MAIN
# ============================================================================
def main():
    client = OpenAI(api_key=OPENROUTER_API_KEY, base_url=OPENROUTER_BASE_URL)

    prompt = (
        "Suggest one creative name for a new pet salon business. The generated name should "
        "evoke positive emotions and the following key features: Professional, friendly, "
        "Personalized Service. Provide only the business name, no explanation."
    )

    print("Generated pet salon names:\n")
    
    # Generate 5 different responses by making 5 separate calls
    for i in range(1, 6):
        resp = client.chat.completions.create(
            model="openai/gpt-4o",
            temperature=0.9,  # High temperature for variety
            max_tokens=50,
            messages=[
                {"role": "system", "content": "You generate short, catchy brand names. Respond with only the business name, no numbering or explanation."},
                {"role": "user", "content": prompt},
            ],
        )
        
        name = resp.choices[0].message.content.strip()
        # Clean up any unwanted formatting
        name = name.replace('"', '').replace("'", '').strip()
        print(f"{i}. {name}")

# ============================================================================
# SCRIPT EXECUTION
# ============================================================================
if __name__ == "__main__":
    main()
```
